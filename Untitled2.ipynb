```python
import pandas as pd

data = {
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Arjun"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": [85, 90, 78, 92, 88]  # Changed strings to integers
}

df = pd.DataFrame(data)

# Calculate average marks
average_marks = df["Marks"].mean()

# Select students who scored above average
top_students = df[df["Marks"] > average_marks]

# Add a new column
df["Performance"] = df["Marks"].apply(
    lambda x: "Excellent" if x >= 90 else "Good"
)

print("Average Marks:", average_marks)
print("\nTop Students:")
print(top_students)

print("\nFinal DataFrame:")
print(df)
```

```python
import pandas as pd

# Original data with string values in 'Marks' column
data = {
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Arjun"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": ["85", "90", "78", "92", "88"]
}

df = pd.DataFrame(data)

# Convert 'Marks' column from string to numeric (integer/float)
df["Marks"] = pd.to_numeric(df["Marks"])

# Calculate average marks (now works because Marks are numbers)
average_marks = df["Marks"].mean()

# Select students who scored above average
top_students = df[df["Marks"] > average_marks]

# Add a new column based on the numeric score
df["Performance"] = df["Marks"].apply(
    lambda x: "Excellent" if x >= 90 else "Good"
)

print("Average Marks:", average_marks)
print("\nTop Students:")
print(top_students)

print("\nFinal DataFrame:")
print(df)
```

```python
import pandas as pd

data = {
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Arjun"],
    "Age": [21, 22, 20, 23, 21],
    "Marks": ["85", "90", "78", "92", "88"]  # These are strings
}

df = pd.DataFrame(data)

# 1. Convert "Marks" column from string to numeric (integer/float)
df["Marks"] = pd.to_numeric(df["Marks"])

# 2. Calculate average marks (now works because Marks are numbers)
average_marks = df["Marks"].mean()

# 3. Select students who scored above average
top_students = df[df["Marks"] > average_marks]

# 4. Add a new column based on the numeric score
df["Performance"] = df["Marks"].apply(
    lambda x: "Excellent" if x >= 90 else "Good"
)

print("Average Marks:", average_marks)
print("\nTop Students:")
print(top_students)

print("\nFinal DataFrame:")
print(df)
```
