```python
import pandas as pd
import numpy as np

data = {
    "Employee": ["Aman", "Riya", "Karan", "Neha", "Vikas", "Simran"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Experience": [2, 4, 3, 5, 1, 6],
    "Performance": [78, 91, 85, 88, 72, 95],
    "Salary": [45000, 62000, 52000, 71000, 40000, 80000]
}

df = pd.DataFrame(data)

def calculate_score(row):
    experience_factor = min(row["Experience"] / 5, 1)
    performance_factor = row["Performance"] / 100

    return (
        experience_factor * 0.35
        + performance_factor * 0.65
    )

df["Score"] = df.apply(calculate_score, axis=1)

department_avg = (
    df.groupby("Department")["Score"]
    .mean()
    .reset_index()
)

df = df.merge(
    department_avg,
    on="Department",
    suffixes=("", "_Average")
)

df["Difference"] = (
    df["Score"] - df["Score_Average"]
)

df["Performance_Level"] = np.where(
    df["Difference"] >= 0,
    "Above Average",
    "Below Average"
)

summary = (
    df.groupby("Department")
    .agg(
        Employees=("Employee", "count"),
        Average_Score=("Score", "mean"),
        Average_Salary=("Salary", "mean"),
        Top_Performer=("Performance", "max")
    )
    .reset_index()
)

summary["Salary_Adjusted"] = (
    summary["Average_Salary"] *
    (1 + summary["Average_Score"] / 100)
)

final_result = summary[
    [
        "Department",
        "Employees",
        "Average_Score",
        "Average_Salary",
        "Top_Performer",
        "Salary_Adjusted"
    ]
]

print(final_result)
```

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman", "Riya", "Karan", "Neha", "Vikas", "Simran",
             "Arjun", "Meera", "Dev", "Isha", "Kabir", "Tara"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance",
                   "IT", "HR", "Finance", "IT", "Finance", "HR"],
    "Experience": [2, 4, 3, 5, 1, 6, 7, 8, 2, 4, 9, 3],
    "Performance": [78, 91, 85, 88, 72, 95, 89, 94, 81, 86, 97, 79],
    "Salary": [45000, 62000, 52000, 71000, 40000, 80000,
               83000, 76000, 59000, 68000, 92000, 51000],
    "Training_Hours": [12, 20, 16, 18, 8, 24, 30, 28, 14, 21, 32, 11],
    "Projects": [3, 2, 4, 3, 1, 5, 7, 4, 2, 6, 8, 3]
}
df = pd.DataFrame(data)

def calculate_score(row):
    experience = min(row["Experience"] / 8, 1)
    performance = row["Performance"] / 100
    training = min(row["Training_Hours"] / 30, 1)
    projects = min(row["Projects"] / 8, 1)
    return experience * 0.25 + performance * 0.5 + training * 0.15 + projects * 0.1

df["Score"] = df.apply(calculate_score, axis=1)
df["Bonus"] = df["Salary"] * (0.04 + df["Score"] * 0.08)
df["Performance_Adjusted"] = df["Score"] * df["Performance"]

summary = (
    df.groupby("Department")
      .agg(
          Average_Score=("Score", "mean"),
          Average_Performance=("Performance", "mean"),
          Average_Salary=("Salary", "mean"),
          Average_Bonus=("Bonus", "mean"),
          Total_Projects=("Projects", "sum")
      )
      .reset_index()
)
summary["Adjusted_Index"] = summary["Average_Score"] * summary["Average_Performance"] / 100
summary["Department_Rank"] = summary["Adjusted_Index"].rank(ascending=False).astype(int)
summary["Salary_Efficiency"] = summary["Average_Performance"] / summary["Average_Salary"]
result = summary.sort_values("Adjusted_Index", ascending=False)
print(result)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Create dataset
data = {
    "age": [21, 24, 27, 30, 33, 36, 39, 42, 45, 48, 51, 54],
    "salary": [22000, 26000, 30000, 35000, 39000, 44000,
               50000, 56000, 62000, 68000, 75000, 82000],
    "experience": [1, 2, 3, 5, 6, 8, 10, 11, 13, 15, 17, 20],
    "promoted": [0, 0, 0, 1, 0, 1, 1, 1, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

# Features and target
X = df[["age", "salary", "experience"]]
y = df["promoted"]

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

# Scale features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train model
model = LogisticRegression(
    random_state=42
)

model.fit(
    X_train_scaled,
    y_train
)

# Evaluate
predictions = model.predict(
    X_test_scaled
)

accuracy = accuracy_score(
    y_test,
    predictions
)

print("Accuracy:", accuracy)

# New employee
new_employee = pd.DataFrame({
    "age": [29],
    "salary": [42000],
    "experience": [5]
})

# Feature engineering
new_employee["salary_per_experience"] = (
    new_employee["salary"] /
    new_employee["experience"]
)

# Prepare prediction data
new_employee = new_employee[
    [
        "age",
        "salary",
        "experience",
        "salary_per_experience"
    ]
]

# Scale new employee data
new_employee_scaled = scaler.transform(
    new_employee
)

# Predict
result = model.predict(
    new_employee_scaled
)

print("Prediction:", result)
```
